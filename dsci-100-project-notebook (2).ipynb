```python
import altair as alt
plot1=alt.Chart(tidied_data).mark_bar().encode(
    x=alt.X('total_wins',bin=True).title('Total Wins'),
    y=alt.Y('count()').title('Frequency')).properties(title='Distribution of Total Match Wins')

plot1
```

```python
plot2=alt.Chart(tidied_data).mark_bar().encode(
    x=alt.X('seed').title('Team Seed'),
    y=alt.Y('count()',stack='normalize').title('Count')).properties(title='Distribution of Tournament Seeds')
plot2
```
